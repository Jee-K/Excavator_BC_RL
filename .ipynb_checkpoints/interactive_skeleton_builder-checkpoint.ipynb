# Interactive Skeleton Builder for URDF

This notebook helps you:
1. Load and visualize your URDF mesh (fully interactive 3D)
2. Explore coordinates by hovering and clicking
3. Define keypoints and joints
4. Create skeleton lines between points
5. Export your skeleton definition

**Instructions:**
- Run cells in order
- **Rotate, zoom, pan** the 3D view with your mouse
- **Hover** over points to see coordinates
- Click points to select them

```python
# Imports
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from scipy.spatial.transform import Rotation as R
import xml.etree.ElementTree as ET
import os

print("✓ Imports loaded")
print("\nTip: Graphs are fully interactive!")
print("  - Left click + drag: Rotate")
print("  - Right click + drag: Pan")
print("  - Scroll: Zoom")
print("  - Hover: See coordinates")
```

## 1. Load URDF and Meshes

```python
# Configuration
urdf_path = "/home/caee/Desktop/Excavator_BC_RL/excavatorURDF/robot_fixed_alternate.urdf"
urdf_dir = os.path.dirname(urdf_path)

# STL loader
def load_stl(filepath):
    """Load binary STL file"""
    with open(filepath, 'rb') as f:
        f.read(80)  # Skip header
        n_triangles = np.frombuffer(f.read(4), dtype=np.uint32)[0]
        
        vertices = []
        for _ in range(n_triangles):
            f.read(12)  # Skip normal
            v1 = np.frombuffer(f.read(12), dtype=np.float32)
            v2 = np.frombuffer(f.read(12), dtype=np.float32)
            v3 = np.frombuffer(f.read(12), dtype=np.float32)
            vertices.append([v1, v2, v3])
            f.read(2)  # Skip attribute
        
        return np.array(vertices)

# Load URDF and meshes
print("Loading URDF and meshes...")
tree = ET.parse(urdf_path)
root = tree.getroot()

link_meshes = {}
for link in root.findall('link'):
    link_name = link.get('name')
    visual = link.find('visual')
    
    if visual is not None:
        geometry = visual.find('geometry')
        if geometry is not None:
            mesh = geometry.find('mesh')
            if mesh is not None:
                mesh_file = mesh.get('filename')
                mesh_path = os.path.join(urdf_dir, mesh_file)
                
                # Get origin transform
                origin = visual.find('origin')
                if origin is not None:
                    xyz = [float(x) for x in origin.get('xyz', '0 0 0').split()]
                    rpy = [float(x) for x in origin.get('rpy', '0 0 0').split()]
                else:
                    xyz = [0, 0, 0]
                    rpy = [0, 0, 0]
                
                if os.path.exists(mesh_path):
                    vertices = load_stl(mesh_path)
                    link_meshes[link_name] = {
                        'vertices': vertices,
                        'origin_xyz': np.array(xyz),
                        'origin_rpy': np.array(rpy),
                        'name': link_name
                    }
                    print(f"  ✓ {link_name}: {vertices.shape[0]} triangles")

print(f"\n✓ Loaded {len(link_meshes)} link meshes")
```

## 2. Interactive 3D Mesh Visualization

**Fully interactive 3D view!**
- Rotate, zoom, pan with your mouse
- Hover over mesh to see coordinates

```python
# Create interactive plot
fig = go.Figure()

# Render each link mesh
for link_name, mesh_data in link_meshes.items():
    vertices = mesh_data['vertices']
    origin_xyz = mesh_data['origin_xyz']
    origin_rpy = mesh_data['origin_rpy']
    
    rot_origin = R.from_euler('xyz', origin_rpy).as_matrix()
    
    # Transform all vertices
    all_vertices = []
    for triangle in vertices:
        for vertex in triangle:
            v = rot_origin @ vertex + origin_xyz
            all_vertices.append(v)
    
    all_vertices = np.array(all_vertices)
    
    # Sample for performance (show ~1000 points per link)
    n_points = len(all_vertices)
    if n_points > 1000:
        step = n_points // 1000
        all_vertices = all_vertices[::step]
    
    # Color based on link type
    if 'part' in link_name.lower():
        color = 'orange'
    elif 'wheel' in link_name.lower() or 'roller' in link_name.lower():
        color = 'gray'
    else:
        color = 'lightblue'
    
    # Add as 3D scatter (mesh3d would be better but more complex)
    fig.add_trace(go.Scatter3d(
        x=all_vertices[:, 0],
        y=all_vertices[:, 1],
        z=all_vertices[:, 2],
        mode='markers',
        marker=dict(
            size=2,
            color=color,
            opacity=0.3
        ),
        name=link_name,
        hovertemplate=f'<b>{link_name}</b><br>' +
                      'X: %{x:.3f}<br>' +
                      'Y: %{y:.3f}<br>' +
                      'Z: %{z:.3f}<br>' +
                      '<extra></extra>'
    ))

# Update layout
fig.update_layout(
    title='Interactive URDF Mesh - Hover to see coordinates',
    scene=dict(
        xaxis_title='X (m)',
        yaxis_title='Y (m)',
        zaxis_title='Z (m)',
        aspectmode='data'
    ),
    width=1200,
    height=800,
    hovermode='closest'
)

print("\n✓ Interactive mesh visualization ready!")
print("\nControls:")
print("  - Left drag: Rotate")
print("  - Right drag: Pan")
print("  - Scroll: Zoom")
print("  - Hover: See coordinates")
print("\nNote the coordinates you want for keypoints!\n")

fig.show()
```

## 3. Show Only Arm Parts

Focus on the excavator arm for easier coordinate finding:

```python
# Create plot with only arm parts
fig_arm = go.Figure()

arm_links = ['part01_pin_1', 'part02_cmpl', 'part03', 'part04']

for link_name in arm_links:
    if link_name not in link_meshes:
        continue
    
    mesh_data = link_meshes[link_name]
    vertices = mesh_data['vertices']
    origin_xyz = mesh_data['origin_xyz']
    origin_rpy = mesh_data['origin_rpy']
    
    rot_origin = R.from_euler('xyz', origin_rpy).as_matrix()
    
    # Transform all vertices
    all_vertices = []
    for triangle in vertices:
        for vertex in triangle:
            v = rot_origin @ vertex + origin_xyz
            all_vertices.append(v)
    
    all_vertices = np.array(all_vertices)
    
    # Sample
    n_points = len(all_vertices)
    if n_points > 2000:
        step = n_points // 2000
        all_vertices = all_vertices[::step]
    
    colors = {'part01_pin_1': 'red', 'part02_cmpl': 'orange', 
              'part03': 'yellow', 'part04': 'green'}
    
    fig_arm.add_trace(go.Scatter3d(
        x=all_vertices[:, 0],
        y=all_vertices[:, 1],
        z=all_vertices[:, 2],
        mode='markers',
        marker=dict(
            size=3,
            color=colors.get(link_name, 'orange'),
            opacity=0.5
        ),
        name=link_name,
        hovertemplate=f'<b>{link_name}</b><br>' +
                      'X: %{x:.3f}<br>' +
                      'Y: %{y:.3f}<br>' +
                      'Z: %{z:.3f}<br>' +
                      '<extra></extra>'
    ))

fig_arm.update_layout(
    title='Excavator Arm Only - Hover to find keypoint coordinates',
    scene=dict(
        xaxis_title='X (m)',
        yaxis_title='Y (m)',
        zaxis_title='Z (m)',
        aspectmode='data'
    ),
    width=1200,
    height=800,
    hovermode='closest'
)

print("\n✓ Arm-only view ready!")
print("\nArm parts:")
print("  🔴 part01_pin_1 - Arm base (where boom connects)")
print("  🟠 part02_cmpl  - Boom")
print("  🟡 part03       - Stick")
print("  🟢 part04       - Bucket")
print("\nHover over points to find good keypoint locations!\n")

fig_arm.show()
```

## 4. Define Keypoints

Based on the coordinates you found, define your keypoints:

```python
# Define your keypoints here
# Format: (name, link_name, offset_xyz)

keypoints_list = [
    # Edit these with your actual coordinates!
    ("arm_base", "part01_pin_1", np.array([0.0, 0.0, 0.0])),
    ("boom_tip", "part02_cmpl", np.array([0.0, 0.0, -1.2])),
    ("stick_tip", "part03", np.array([0.0, 0.0, -0.8])),
    ("bucket_base", "part04", np.array([0.0, 0.0, 0.0])),
    ("bucket_floor", "part04", np.array([0.0, -0.15, -0.2])),
    ("bucket_tip", "part04", np.array([0.0, 0.0, -0.4])),
]

# Convert to dictionary
keypoints = {name: (link, offset) for name, link, offset in keypoints_list}

print("Defined keypoints:")
for i, (name, link, offset) in enumerate(keypoints_list):
    print(f"  {i}. {name:20s} on '{link}' at [{offset[0]:.3f}, {offset[1]:.3f}, {offset[2]:.3f}]")
```

## 5. Define Skeleton Lines

Connect keypoints to form your skeleton:

```python
# Define skeleton connections
skeleton_lines = [
    ("arm_base", "boom_tip"),
    ("boom_tip", "stick_tip"),
    ("stick_tip", "bucket_base"),
    ("bucket_base", "bucket_floor"),
    ("bucket_floor", "bucket_tip"),
    ("bucket_tip", "bucket_base"),  # Close triangle
]

print("Skeleton connections:")
for i, (p1, p2) in enumerate(skeleton_lines):
    print(f"  {i}. {p1} → {p2}")
```

## 6. Visualize Complete Skeleton

See your skeleton overlaid on the mesh!

```python
# Create figure with mesh + skeleton
fig_complete = go.Figure()

# Add arm meshes (lighter)
for link_name in arm_links:
    if link_name not in link_meshes:
        continue
    
    mesh_data = link_meshes[link_name]
    vertices = mesh_data['vertices']
    origin_xyz = mesh_data['origin_xyz']
    origin_rpy = mesh_data['origin_rpy']
    
    rot_origin = R.from_euler('xyz', origin_rpy).as_matrix()
    all_vertices = []
    for triangle in vertices:
        for vertex in triangle:
            v = rot_origin @ vertex + origin_xyz
            all_vertices.append(v)
    
    all_vertices = np.array(all_vertices)
    if len(all_vertices) > 1000:
        step = len(all_vertices) // 1000
        all_vertices = all_vertices[::step]
    
    fig_complete.add_trace(go.Scatter3d(
        x=all_vertices[:, 0],
        y=all_vertices[:, 1],
        z=all_vertices[:, 2],
        mode='markers',
        marker=dict(size=2, color='lightgray', opacity=0.2),
        name=link_name,
        showlegend=False,
        hoverinfo='skip'
    ))

# Compute keypoint positions (approximate - using link origins + offsets)
keypoint_positions = {}
for name, (link, offset) in keypoints.items():
    if link in link_meshes:
        origin_xyz = link_meshes[link]['origin_xyz']
        origin_rpy = link_meshes[link]['origin_rpy']
        rot_origin = R.from_euler('xyz', origin_rpy).as_matrix()
        pos = rot_origin @ offset + origin_xyz
        keypoint_positions[name] = pos

# Plot keypoints
kp_names = list(keypoint_positions.keys())
kp_coords = np.array([keypoint_positions[name] for name in kp_names])

fig_complete.add_trace(go.Scatter3d(
    x=kp_coords[:, 0],
    y=kp_coords[:, 1],
    z=kp_coords[:, 2],
    mode='markers+text',
    marker=dict(size=10, color='red', symbol='circle', 
                line=dict(width=2, color='black')),
    text=kp_names,
    textposition='top center',
    textfont=dict(size=10, color='darkred'),
    name='Keypoints',
    hovertemplate='<b>%{text}</b><br>' +
                  'X: %{x:.3f}<br>' +
                  'Y: %{y:.3f}<br>' +
                  'Z: %{z:.3f}<br>' +
                  '<extra></extra>'
))

# Draw skeleton lines
for p1_name, p2_name in skeleton_lines:
    if p1_name in keypoint_positions and p2_name in keypoint_positions:
        p1 = keypoint_positions[p1_name]
        p2 = keypoint_positions[p2_name]
        
        fig_complete.add_trace(go.Scatter3d(
            x=[p1[0], p2[0]],
            y=[p1[1], p2[1]],
            z=[p1[2], p2[2]],
            mode='lines',
            line=dict(color='blue', width=6),
            name=f'{p1_name}-{p2_name}',
            showlegend=False,
            hoverinfo='skip'
        ))

fig_complete.update_layout(
    title='Complete Skeleton - Keypoints and Lines',
    scene=dict(
        xaxis_title='X (m)',
        yaxis_title='Y (m)',
        zaxis_title='Z (m)',
        aspectmode='data'
    ),
    width=1200,
    height=800
)

print("\n✓ Skeleton visualization complete!")
print("\n🔴 Red dots = Keypoints")
print("🔵 Blue lines = Skeleton connections\n")

fig_complete.show()
```
